__Imports__

```python
import numpy as np
import matplotlib.pyplot as plt
```

__Problem 1a__: Plotting log-log diagram for u_coil/u_ground

```python
w = np.logspace(0, 4, 1000)
w_0 = 100
u_ratio = np.abs(1 / (1 - w**2/w_0**2))
H_low = np.ones_like(w)
H_high = (w_0**2) / (w**2)

plt.title("log-log diagram av forholdet mellom u_coil og u_ground")
plt.plot(w, u_ratio, label='overføringsfunksjon')
plt.plot(w, H_low, '--')
plt.plot(w, H_high, '--')
plt.grid()
plt.legend()

plt.xscale('log')
plt.xlabel("w")
plt.yscale('log')
plt.ylabel(r'$|u_{coil}/u_{ground}|$')

plt.show()

```

__Problem 1c__: ratio e/u_ground

```python
B = 1
l = 1
w = np.logspace(0, 4, 1000)
w_0 = 100

ratio_c = np.abs(B * l * (1 / (1 - w**2/w_0**2) - 1))

plt.title("log-log diagram for e/u_ground")
plt.plot(w, ratio_c, label="|e/u_ground|")
plt.plot(w, H_low)
plt.plot(w, H_high)
plt.grid()
plt.legend()

plt.xscale('log')
plt.xlabel("w")
plt.yscale('log')
plt.ylabel(r"e/u_{ground}")

plt.show()
```

__Problem 2b__: script for calculating R's

```python
import numpy as np

A=3.9083*10e-3
B=5.775*10e-7
R0=100

T = [0, 100, 500]
R_approx = []
R_true = []
D=[]

for t in T:
    D_calculate = A + B * t
    D.append(D_calculate)
    R_approx.append(R0 * (1 + D_calculate*t))
    R_true.append(R0 * (1 + A*t + B*t**2))

print(f"D: {D} \nR_approx: {R_approx} \nR_true: {R_true}")
```
